## 1. Import Required Libraries
Import essential libraries for data manipulation and visualization.

```python
# Import essential libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## 2. Simple Data Manipulation Task
Create a small DataFrame and perform basic operations like filtering, sorting, and adding a new column.

```python
# Create a small DataFrame

data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva'],
    'Score': [85, 92, 78, 90, 88],
    'Attempts': [1, 2, 3, 2, 1]
}
df = pd.DataFrame(data)

# Filter: Students with Score > 85
filtered_df = df[df['Score'] > 85]

# Sort by Score descending
sorted_df = df.sort_values(by='Score', ascending=False)

# Add a new column: Passed (True if Score >= 80)
df['Passed'] = df['Score'] >= 80

df, filtered_df, sorted_df
```

## 3. Basic Data Visualization
Plot a simple bar chart to visualize the scores of students.

```python
# Bar plot of student scores
plt.figure(figsize=(6,4))
plt.bar(df['Name'], df['Score'], color='skyblue')
plt.xlabel('Name')
plt.ylabel('Score')
plt.title('Student Scores')
plt.show()
```

## 4. Write and Run a Function
Implement a function to calculate the mean score and demonstrate its usage.

```python
# Function to calculate mean score
def mean_score(df):
    return df['Score'].mean()

# Demonstrate usage
mean = mean_score(df)
print(f"Mean Score: {mean}")
```

## 5. Unit Test the Function
Write and execute a basic unit test to verify the correctness of the implemented function.

```python
# Unit test for mean_score function
def test_mean_score():
    test_df = pd.DataFrame({'Score': [10, 20, 30]})
    assert mean_score(test_df) == 20, "Test failed: mean should be 20"
    print("Test passed!")

test_mean_score()
```

```python
# Text generation using Hugging Face Transformers (faster demo)
def generate_text(prompt, max_length=20):
    from transformers import pipeline
    # Use a tiny model for speed
    generator = pipeline('text-generation', model='sshleifer/tiny-gpt2')
    result = generator(prompt, max_length=max_length, num_return_sequences=1)
    return result[0]['generated_text']

# Example usage
prompt = "Once upon a time, in a world of AI,"
generated = generate_text(prompt)
print(generated)
```
