```python
# Distribution of labels: bar chart and pie chart side by side
import matplotlib.pyplot as plt
label_map = {0:'No object', 2:'Bat', 1:'Ball', 3:'Wicket'}
label_counts = df['label'].value_counts().sort_index()
label_counts.index = label_counts.index.map(label_map)
label_percentage = df['label'].value_counts(normalize=True).sort_index() * 100
label_percentage.index = label_percentage.index.map(label_map)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))
# Bar chart
label_counts.plot(kind='bar', ax=axes[0], color='skyblue')
axes[0].set_xlabel('Label')
axes[0].set_ylabel('Count')
axes[0].set_title('Distribution of Labels in Annotations')
axes[0].grid(axis='y', alpha=0.3)
# Pie chart
label_percentage.plot(kind='pie', ax=axes[1], autopct='%1.1f%%', colors=['#a6cee3','#1f78b4','#b2df8a','#33a02c'])
axes[1].set_title('Percentage Distribution of Labels')
axes[1].set_ylabel('')  # Hide y-label for pie chart
plt.tight_layout()
plt.show()
```
