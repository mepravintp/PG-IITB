# Visualize Model Predictions on Images
This notebook reads the final predictions CSV, overlays color-coded cell grids on each image, and saves the processed images for inspection.

```python
# Section 1: Import Required Libraries
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from skimage import io
```

```python
# Section 2: Read Predictions CSV
csv_path = r"C:\Users\pravi\PG IITB\cricket_object_detection\outputs\final_predictions_test.csv"
pred_df = pd.read_csv(csv_path)
print("Length of pred_df:", len(pred_df))
cell_cols = [f"c{str(i+1).zfill(2)}" for i in range(64)]

print("Number of unique images:", len(pred_df["ImageFileName"].unique()))
```

```python
# Section 3: Load Images and Process Grid Labels
image_dir = r"C:\Users\pravi\PG IITB\cricket_object_detection\data\train_modified"
output_dir = r"C:\Users\pravi\PG IITB\cricket_object_detection\outputs\processed_test"
os.makedirs(output_dir, exist_ok=True)
cell_height, cell_width = 75, 100  # For 800x600 images, 8x8 grid
```

```python
# Section 4: Visualize and Color-Code Cell Grids
color_map = {0: 'gray', 1: 'red', 2: 'green', 3: 'blue'}

from matplotlib.patches import Rectangle

def visualize_and_save(img_name, cell_labels):
    img_path = os.path.join(image_dir, img_name)
    if not os.path.exists(img_path):
        print(f"Image not found: {img_name}")
        return
    img = io.imread(img_path)
    fig, ax = plt.subplots(figsize=(8, 6))
    ax.imshow(img)
    for idx, label in enumerate(cell_labels):
        r, c = divmod(idx, 8)
        y1, y2 = r * cell_height, (r + 1) * cell_height
        x1, x2 = c * cell_width, (c + 1) * cell_width
        # Draw cell border
        rect = Rectangle((x1, y1), cell_width, cell_height, fill=False, edgecolor=color_map[label], linewidth=2)
        ax.add_patch(rect)
        # Draw darker background highlight for labels 1,2,3
        if label in [1, 2, 3]:
            highlight = Rectangle((x1, y1), cell_width, cell_height, fill=True, facecolor=color_map[label], alpha=0.5, edgecolor=None)
            ax.add_patch(highlight)
    ax.set_title(img_name)
    ax.axis('off')
    plt.tight_layout()
    # Save image
    out_path = os.path.join(output_dir, img_name)
    fig.savefig(out_path)
    plt.close(fig)
```

```python
# Section 5: Save Processed Images
for idx, row in pred_df.iterrows():
    img_name = row['ImageFileName']
    print(f"Processing image: {img_name} , {idx+1}/{len(pred_df)}")
    cell_labels = [row[col] for col in cell_cols]
    visualize_and_save(img_name, cell_labels)
print(f"Processed images saved to {output_dir}")
```
