## Section 1: CNN Fundamentals & Architecture

### Q1: Basic CNN Components
**Question:** A CNN layer gets an input of shape (1, 32, 32, 3). What do these dimensions represent?

**Answer:** 
- 1: Batch size (1 image)
- 32×32: Height and width of image
- 3: Number of channels (RGB)

---

### Q2: Convolution Operation
**Question:** If you apply a 3×3 filter to a 5×5 image with stride=1 and padding=0, what will be the output dimensions?

**Answer:** 
Using formula: Output_size = (Input_size - Filter_size) / Stride + 1
- Output_size = (5 - 3) / 1 + 1 = 3
- Output dimensions: **3×3**

---

### Q3: Padding Types
**Question:** Compare 'same' padding and 'valid' padding in convolution operations.

**Answer:**
- **Valid Padding**: No padding, output size shrinks
  - Output = (Input - Filter) / Stride + 1
- **Same Padding**: Pads input to keep output size = input size (with stride=1)
  - Padding needed = (Filter_size - 1) / 2
  - Best for maintaining spatial dimensions

---

### Q4: Pooling vs Convolution
**Question:** What's the difference between max pooling and average pooling?

**Answer:**
- **Max Pooling**: Takes maximum value from each window
  - Preserves sharp features, reduces noise
  - Better for feature detection
- **Average Pooling**: Takes mean of all values in window
  - Smooths out features
  - Better for reducing dimensionality

## Section 3: Feature Maps & Filters

### Q8: What Does a Filter Learn?
**Question:** Explain what convolutional filters learn in different layers of a CNN.

**Answer:**
- **Early layers**: Low-level features (edges, corners, textures)
- **Middle layers**: Mid-level features (shapes, patterns)
- **Deep layers**: High-level features (objects, faces, specific patterns)
- This is due to hierarchical feature extraction

---

### Q9: Receptive Field
**Question:** What is receptive field and why is it important?

**Answer:**
- **Receptive Field**: The region of input that a neuron can see
- **Importance**:
  - Larger receptive field → captures larger context
  - Can be increased by:
    - Larger kernel size
    - Multiple stacked layers
    - Dilation
    - Stride > 1

---

### Q10: Why Convolution Over Fully Connected?
**Question:** Why are convolutional layers better than fully connected layers for image processing?

**Answer:**
- **Parameter Sharing**: Same filter used across spatial locations → fewer parameters
- **Spatial Invariance**: Detects features regardless of position
- **Local Connectivity**: Captures local patterns efficiently
- **Translation Equivariance**: Shifted input → shifted output
- **Memory Efficiency**: Much fewer parameters than FC layers

## Section 4: Backpropagation & Training

### Q11: Gradient Flow in CNNs
**Question:** Explain how gradients flow during backpropagation and what is "vanishing gradient problem"?

**Answer:**
- **Gradient Flow**: During backprop, gradients are computed layer by layer moving backwards
- **Vanishing Gradient**: 
  - In deep networks, gradients become very small (multiply chain of factors < 1)
  - Early layers get almost zero gradients → no learning
- **Solutions**:
  - Batch Normalization
  - ReLU activation (gradient = 1 for positive x)
  - Skip connections (ResNet)
  - Careful weight initialization

---

### Q12: Batch Normalization in CNNs
**Question:** What does Batch Normalization do and where should it be placed?

**Answer:**
- **Function**: Normalizes inputs to each layer to have mean ≈ 0 and std ≈ 1
- **Benefits**:
  - Stabilizes training
  - Allows higher learning rates
  - Acts as regularization
  - Reduces internal covariate shift
- **Placement**: Usually after Conv layer, before activation
  - Conv → BatchNorm → Activation

---

### Q13: Dropout Regularization
**Question:** How does dropout prevent overfitting? What's the typical dropout rate?

**Answer:**
- **Mechanism**: Randomly deactivates neurons during training with probability p
- **Effect**: Forces network to learn redundant representations
- **Prevents**: Co-adaptation of neurons, overfitting
- **Typical rates**: 
  - Early layers: 0.2 (20%)
  - Deep layers: 0.5 (50%)
  - Higher than 0.5 usually harmful
- **Note**: Disabled during inference, activations scaled during training

### Q15: VGG Architecture
**Question:** What principle did VGG networks follow? Why use 3×3 filters instead of larger ones?

**Answer:**
- **Principle**: "Simplicity is key" - stack small 3×3 filters
- **Advantages of 3×3**:
  - Two 3×3 = receptive field of 5×5 but fewer parameters
  - Three 3×3 = receptive field of 7×7 but fewer parameters
  - More non-linearity between layers
- **Formula**: (3×3)^n has fewer parameters than k×k where k = 2n+1

---

### Q16: ResNet Skip Connections
**Question:** Why do ResNets use skip connections? How do they solve the degradation problem?

**Answer:**
- **Problem**: Deeper networks perform worse (degradation problem)
  - Not due to overfitting (training error also increases)
  - Optimization becomes harder
- **Solution - Skip Connection**:
  - y = F(x) + x (instead of y = F(x))
  - Allows identity mapping when F(x) ≈ 0
  - Gradient flow: ∂y/∂x = ∂F/∂x + 1 (always ≥ 1)
  - Greatly improves backpropagation in very deep networks

---

### Q17: Inception Module (GoogLeNet)
**Question:** What problem did the Inception module solve?

**Answer:**
- **Problem**: What's the best filter size? 1×1, 3×3, 5×5, or pooling?
- **Solution**: Use all of them in parallel!
- **Design**:
  - Parallel 1×1, 3×3, 5×5 convolutions
  - Plus max pooling
  - Concatenate outputs
- **Benefit**: 
  - Captures multi-scale features
  - 1×1 convs reduce parameters (dimensionality reduction)
  - More efficient than large filters

### Q20: Loss Functions for Classification
**Question:** When would you use Categorical Crossentropy vs Binary Crossentropy?

**Answer:**
- **Binary Crossentropy**:
  - Binary classification (2 classes)
  - Sigmoid activation in last layer
  - Loss = -[y*log(ŷ) + (1-y)*log(1-ŷ)]
- **Categorical Crossentropy**:
  - Multi-class classification (>2 classes)
  - Softmax activation in last layer
  - Loss = -Σ y_i * log(ŷ_i)

## Section 7: Advanced Concepts

### Q21: 1×1 Convolution
**Question:** What's the purpose of 1×1 convolution? Where is it commonly used?

**Answer:**
- **Purpose**:
  - Dimensionality reduction (reduce channels)
  - Adds non-linearity
  - Reduces parameters
  - Cross-channel interactions
- **Common Uses**:
  - Inception modules (before 3×3, 5×5)
  - ResNet bottleneck blocks
  - Channel compression
- **Example**: 64 channels → 1×1 conv → 16 channels

---

### Q22: Dilated/Atrous Convolution
**Question:** What is dilated convolution? Why use it?

**Answer:**
- **Definition**: Filter with gaps (dilation rate > 1)
  - Dilation=1: Normal convolution
  - Dilation=2: Every other element
- **Benefits**:
  - Larger receptive field without more parameters
  - Preserves resolution better than pooling
  - No information loss from skipped pixels
- **Common Uses**: Semantic segmentation, dense prediction tasks

---

### Q23: Depthwise Separable Convolution
**Question:** How does depthwise separable convolution reduce parameters?

**Answer:**
- **Standard Conv**: Input_channels × Output_channels × Kernel_size²
- **Depthwise Separable**:
  - Depthwise: Applied per channel (Input_channels × 1 × Kernel_size²)
  - Pointwise: 1×1 conv (Input_channels × Output_channels × 1)
- **Parameter Reduction**:
  - Example: 3×3, 256→512 channels
  - Standard: 256 × 512 × 9 = 1,179,648
  - Depthwise sep: 256×9 + 256×512 = 133,120
  - ~8.8x fewer parameters!
- **Used in**: MobileNet, EfficientNet (lightweight models)

## Section 8: Performance & Testing

### Q24: Evaluation Metrics for Image Classification
**Question:** Explain Precision, Recall, F1-Score, and when to use each.

**Answer:**
- **Accuracy**: % correct predictions
  - Use: Balanced datasets
- **Precision**: % of positive predictions that were correct
  - Precision = TP / (TP + FP)
  - Use: When false positives are costly
- **Recall**: % of actual positives that were found
  - Recall = TP / (TP + FN)
  - Use: When false negatives are costly
- **F1-Score**: Harmonic mean of Precision & Recall
  - F1 = 2 * (Precision * Recall) / (Precision + Recall)
  - Use: Imbalanced datasets, want balance

---

### Q25: Confusion Matrix & ROC-AUC
**Question:** What information does a confusion matrix provide?

**Answer:**
- **Confusion Matrix** (for binary classification):
  ```
                 Predicted Positive  Predicted Negative
  Actual Positive        TP                  FN
  Actual Negative        FP                  TN
  ```
- **Derived Metrics**:
  - Sensitivity (Recall) = TP / (TP + FN)
  - Specificity = TN / (TN + FP)
  - Precision = TP / (TP + FP)
  - Accuracy = (TP + TN) / Total
- **ROC-AUC**: Plot TPR vs FPR at different thresholds
  - AUC = 1: Perfect classifier
  - AUC = 0.5: Random guessing

## Section 9: Real-World Applications & Challenges

### Q26: Object Detection with CNNs
**Question:** How do you modify a CNN for object detection instead of classification?

**Answer:**
- **Main Differences**:
  - Need to predict: class AND bounding box (x, y, width, height)
  - Multiple objects in single image
- **Approaches**:
  - **R-CNN**: Propose regions first, classify each
  - **YOLO**: Single shot - predict boxes directly
  - **SSD**: Multi-scale predictions
- **Loss Function**: Combined classification + localization loss
  - Loss = Classification_loss + λ * Localization_loss

---

### Q27: Semantic Segmentation vs Instance Segmentation
**Question:** What's the difference and how are CNNs adapted?

**Answer:**
- **Semantic Segmentation**: Pixel-level classification
  - Output: Class label for each pixel
  - Doesn't distinguish between instances
- **Instance Segmentation**: Pixel classification + instance differentiation
  - Output: Class + instance ID for each pixel
- **CNN Architectures**:
  - U-Net: Encoder-decoder with skip connections
  - Fully Convolutional Networks (FCN)
  - Mask R-CNN: R-CNN + segmentation head

---

### Q28: Class Imbalance Problem
**Question:** What's class imbalance and how do you handle it?

**Answer:**
- **Problem**: One class has many more examples than others
  - Example: 95% negative, 5% positive
  - Model learns to always predict majority class
- **Solutions**:
  - **Data**: Oversampling minority, undersampling majority
  - **Loss**: Class weights (penalize misclassifying minority)
  - **Metrics**: Use F1, Precision-Recall instead of accuracy
  - **Threshold Tuning**: Adjust decision boundary
- **Cost-sensitive Learning**: Loss = Σ class_weight_i * loss_i

## Section 10: Computational Efficiency & Deployment

### Q29: Model Compression Techniques
**Question:** How do you reduce model size for deployment on mobile devices?

**Answer:**
- **Quantization**: Reduce precision
  - 32-bit float → 8-bit integer
  - ~4x smaller, minimal accuracy loss
- **Pruning**: Remove unimportant weights
  - Set small weights to zero
  - Can achieve 90% sparsity
- **Knowledge Distillation**: Teacher-student training
  - Large teacher network → small student network
- **Architecture Search**: Find efficient architectures
  - MobileNet, EfficientNet, SqueezeNet

### Q30: Inference Optimization
**Question:** How can you speed up CNN inference?

**Answer:**
- **Hardware**: Use GPU/TPU for parallel computation
- **Batch Processing**: Process multiple images together
- **Model Simplification**:
  - Remove dropout (only for training)
  - Fuse BatchNorm into Conv weights
  - Use simpler activations
- **Frameworks**: TensorRT, ONNX, CoreML for deployment
- **Caching**: Reuse computed features when possible

````python
# Practice: Calculate output dimensions
def calculate_conv_output(input_size, kernel_size, stride=1, padding=0):
    """
    Calculate output size after convolution
    Formula: Output = (Input - Kernel + 2*Padding) / Stride + 1
    """
    output = (input_size - kernel_size + 2*padding) // stride + 1
    return output

# Test cases
print("Test 1: 5×5 input, 3×3 kernel, stride=1, padding=0")
print(f"Output: {calculate_conv_output(5, 3, stride=1, padding=0)}×{calculate_conv_output(5, 3, stride=1, padding=0)}")
print(f"Expected: 3×3\n")

print("Test 2: 224×224 input, 3×3 kernel, stride=1, padding=1")
print(f"Output: {calculate_conv_output(224, 3, stride=1, padding=1)}×{calculate_conv_output(224, 3, stride=1, padding=1)}")
print(f"Expected: 224×224 (same padding)\n")

print("Test 3: 28×28 input, 2×2 kernel, stride=2, padding=0")
print(f"Output: {calculate_conv_output(28, 2, stride=2, padding=0)}×{calculate_conv_output(28, 2, stride=2, padding=0)}")
print(f"Expected: 14×14 (max pooling)")
````

````python
# Practice: Calculate number of parameters
def count_conv_parameters(in_channels, out_channels, kernel_h, kernel_w, bias=True):
    """
    Count parameters in a convolutional layer
    Weights: kernel_h × kernel_w × in_channels × out_channels
    Bias: out_channels (one per filter)
    """
    weights = kernel_h * kernel_w * in_channels * out_channels
    bias_params = out_channels if bias else 0
    total = weights + bias_params
    return weights, bias_params, total

# Example: Calculate for VGG-like layer
print("Conv Layer: 64 input channels, 128 output channels, 3×3 kernel")
weights, bias, total = count_conv_parameters(64, 128, 3, 3)
print(f"Weights: {weights:,}")
print(f"Bias: {bias:,}")
print(f"Total Parameters: {total:,}")
print(f"Expected: 204,928\n")

print("1×1 Convolution: 256 input channels, 64 output channels")
weights, bias, total = count_conv_parameters(256, 64, 1, 1)
print(f"Total Parameters: {total:,}")
print(f"Expected: 16,448")
````
